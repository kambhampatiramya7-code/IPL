```
from sentence_transformers import SentenceTransformer
import faiss
import numpy as np
```

```
documents = [
    "Genenrative AI creates new content.",
    "RAG combines retrival and generation.",
    "Kibana is used for data visualization.",
    "Elasticsearch stores data efficiently."
]
```

```
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```
doc_vectors = model.encode(documents)
```

```
index = faiss.IndexFlatL2(doc_vectors.shape[1])
index.add(np.array(doc_vectors))
```

```
query = "What is Generative AI?"
query_vector = model.encode([query])
```

```
D, I = index.search(np.array(query_vector), 1)
print("Answer:", documents[I[0][0]])
```

```
#Iterative Prompt Refinement
#Task: Summarise Romeo and Juliet


#Iteration 1: Minimal Prompt
prompt1 = "Summarise the plot of Romeo and Juliet."

output1 = "Romeo and Juliet are two young lovers from families who fall in love and secretly marry. Their tragic deaths finally end of the conflict between their families."

print("ITERATION 1")
print("prompt:", prompt1)
print("Output:", output1)

#Iteration 2: Add length and style
prompt2 = "Summarise Romeo and Juliet exactly two sentences using a concise and formal style."

output2 = "Romeo and Juliet are two young lovers from the feuding Montague and Capulet families who secretly marry despite their families' hatred. A series of tragic events leads to their deaths finally brings peace between their families."

print("ITERATION 2")
print("prompt:", prompt2)
print("Output:", output2)

#Iteration 3: Add Key Content
prompt3 = "Summarise Romeo and Juliet in exactly two sentences. Include the setting, conflict, and theme."

output3 = "Set in Verona, Romeo and Juliet tells the story of two young lovers from the rival Montague and Capulet families who secretly marry. Their tragic deaths highlight the themes of love, fate, and the destructive effects of family conflict."


print("ITERATION 3")
print("prompt:", prompt3)
print("Output:", output3)
```

```
#Import the Required Library
from sentence_transformers import SentenceTransformer
```

```
#Load the Pre-trained Model
model = SentenceTransformer('all-MiniLM-L6-v2')
```

```
#Define input Sentences
sentences = [
    "Machine learing is a branch of artificial intelligence.",
    "Artificial intelligence helps machines learn from data.",
    "I like to play cricket."
]
```

```
#Generate Embeddings
embeddings = model.encode(sentences)
```

```
#Display the Embeddings
for i, embedding in enumerate(embeddings):
  print("Sentence:", sentences[i])
  print("Embedding:", embedding)
  print()
```

```
#Check the Embedding Dimension
print("Number of sentence:", len(embeddings))
print("Embedding domension:", len(embeddings[0]))
```
