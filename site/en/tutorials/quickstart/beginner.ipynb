```
from transformers import AutoTokenizer, AutoModelForCausalLM
#Load tokenizer and model
tokenizer = AutoTokenizer.from_pretrained("distilgpt2")
model = AutoModelForCausalLM.from_pretrained("distilgpt2")

prompt = "Artificial intelligence helps humans by"

inputs = tokenizer(prompt, return_tensors="pt")

outputs = model.generate(
    inputs["input_ids"],
    max_length=40
)

#Convert tokens back to text
generated_text = tokenizer.decode(outputs[0], skip_special_tokens=True)
print(generated_text)
```

```
from sentence_transformers import SentenceTransformer
import faiss
import numpy as np
```

```
text ="""
Artificial Intelligence is the field of computer science that develops system capable of performing tasks that normally require human intelligence.

Machine Learning is a subset of Artificial Intelligence. It allows computers to learn patterns from data and make predictions without being explicity programmed.

Deep Learning is a subset of Machine Learning that uses neural networks with multiple layers.

Natural Language Processing is a branch of Artificial Intelligence that enables computers to understand, process and generate human language.

Generative AI is a type of artificial intelligence that can generate new content such as text, images, audio and code.
"""

print(text)
```

```
chunks = text.split("\n\n")

chunks = [chunk.strip() for chunk in chunks if chunk.strip()]

print("Number of chunks:", len(chunks))

for i, chunk in enumerate(chunks):
    print("\nChunk", i)
    print(chunk)
```

```
model = SentenceTransformer("all-MiniLM-L6-v2")

embeddings = model.encode(chunks)

print("Embedding shape:", embeddings.shape)
```

```
dimension = embeddings.shape[1]

index = faiss.IndexFlatL2(dimension)

index.add(np.array(embeddings).astype("float32"))

print("Documents stored in FAISS:", index.ntotal)
```

```
def retrieve(query, k=2):
  query_embedding = model.encode([query])

  distances, indices = index.search(
      np.array(query_embedding).astype("float32"),
      k
  )

  results = []

  for i in indices[0]:
      results.append(chunks[i])

  return results
```

```
query = "What is Machine Learning?"

results = retrieve(query)

print("Retrieved Information:\n")

for result in results:
    print(result)
```
